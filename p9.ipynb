```python
import pandas as pd

def csv_to_json(csv_file: str, json_file: str):
    """
    Convierte un archivo CSV en un archivo JSON.

    :param csv_file: Nombre del archivo CSV de entrada.
    :param json_file: Nombre del archivo JSON de salida.
    """
    try:
        # Leer el CSV
        df = pd.read_csv(csv_file)

        # Convertir a JSON
        df.to_json(json_file, orient="records", lines=False, force_ascii=False, indent=4)

        print(f"✅ Archivo convertido con éxito: {json_file}")
    except Exception as e:
        print(f"❌ Error: {e}")


if __name__ == "__main__":
    csv_to_json("articles_data_updated.csv", "articles_data_updated.json")
```
